### Naming Conventions
A name with more than one word cannot have spaces. There are 3 common styles:
- **Snake case:** `has_money` (the style we use in Python)

- **Pascal case:** `HasMoney`

````python
has_money = True   # Snake case
hasMoney = True    # Camel case
HasMoney = True    # Pascal case

print(has_money, hasMoney, HasMoney)
````

## 6. Conditions
Conditions let the code choose a path.
- `if`: if the condition is true, run this
- `elif`: otherwise, if this condition is true, run this
- `else`: otherwise, run this

Note: the line under the condition must be **indented** (pushed to the right).

````python
# Default Approach
money = 500
if money > 1000:
    print("You are rich")
elif money > 0:
    print("You have some money")
else:
    print("You are BROKE!")
````

### Ternary Operator
A short way to write if/else in **one line**:

`value_if_true if condition else value_if_false`

````python
# Ternary Operator
age = 20
message = "u can drive a car" if age >= 18 else "u cannot drive a car"
print(message)
````

````python
age = 20
if age >= 18:
    message = "u can drive a car"
else:
    message = "u cannot drive a car"
print(message)
````

## 7. Logical Operators
We use them to join more than one condition:
- `and`: **both** must be true
- `or`: **at least one** must be true
- `not`: flips the value (`True` becomes `False`)

````python
print(True and True)
print(True and False)
print(True or False)
print(False or False)
print(not True)
````

````python
is_hungry = True
has_money = True
restaurant_open = True
meeting_friends = False

if (is_hungry and has_money) or (meeting_friends and restaurant_open):
    print("i want food")
else:
    print("i don't want food")
````

## 8. While Loop
A while loop repeats code **as long as the condition is true**. When the condition becomes false, it stops.

```
while condition:
    code that repeats
```

Warning: we must change the value inside the loop, or it will run forever.

````python
# While Loop
number = 1

while number <= 5:
    print(number)
    number += 1
print("Done!")
````

````python
number = 1
while number < 6:
    print(number)
    number += 1
print("Done!")
````

````python
number = 1
while number < 6:
    print(number)
    number += 2
print("Done!")
````

````python
number = 5
while number > 0:
    print(number)
    number -= 1
print("Done!")
````

````python
number = 2
while number <= 10:
    print(number)
    number += 2
````

## 9. For Loop
A for loop repeats code **a known number of times**. `range()` makes the numbers for us.

```
for number in range(...):
    code that repeats
```

Difference from `while`: we do not need `number = 1` or `number += 1`. The `for` loop does it for us.

`range(start, stop, step)` means: begin at `start`, stop **before** `stop`, and jump by `step`.

````python
for number in range(5):
    print(number)
print("Done!")
````

````python
for number in range(1, 6):
    print(number)
print("Done!")
````

````python
for number in range(1, 6, 2):
    print(number)
print("Done!")
````

````python
for number in range(5, 0, -1):
    print(number)
print("Done!")
````

````python
# while
number = 1
while number <= 5:
    print(number)
    number += 1

# for
for number in range(1, 6):
    print(number)
````

## 10. Nested Loops
A loop inside another loop. The **inner loop finishes completely** every time the outer loop moves one step.

Like a clock: the minute hand makes a full circle (60 minutes) each time the hour hand moves one step.

````python
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i} x {j} = {i * j}")
print("Done!")
````

````python
i = 1
while i < 4:
    j = 1          # j goes back to 1 every time the outer loop repeats
    while j < 4:
        print(f"{i} x {j} = {i * j}")
        j += 1
    i += 1
print("Done!")
````

## 11. Functions
A function is a **named block of code**. We write it once and can use it (call it) many times.
- `def` creates a function
- The names inside the brackets are **parameters** (the values going in)
- `return` sends a result back

Note: defining a function does nothing by itself. We must **call** it.

````python
print("Hi")
print("Hi")
print("Hi")
````

````python
def say_hi():
    print("Hi")

say_hi()
say_hi()
say_hi()
````

````python
def say(word):
    print(word)

say("Hi")
say("Bye")
say("Welcome")
````

````python
def add_ten(number):
    return number + 10

result = add_ten(5)
print(result)
````

````python
def salary(bonus):
    return 4000 + bonus

print(salary(500))
````

````python
def salary(bonus, deduction):
    return 4000 + bonus - deduction

print(salary(500, 100))
````

## 13. Looping Through Lists
Instead of printing each item by hand (`grades[0]`, `grades[1]`, ...), we use `for` to go through all the items automatically.

````python
grades = [85, 90, 78, 92, 88, 95, 80, 87, 91, 89]

print(grades[0])
print(grades[1])
print(grades[2])
````

````python
grades = [85, 90, 78, 92, 88, 95, 80, 87, 91, 89]

for i in range(len(grades)):
    print(grades[i])
````

## 15. String vs List
A string works like a list: every character has an index that starts at 0.

We can also cut a part of it using **slicing**: `[start:stop]` (it stops **before** `stop`).

````python
name = "Ahmed"
print(name[0])
print(name[-1])
````

````python
print(name[0:3])
print(name)
````

````python
grades = [10, 20, 30, 40, 50]
print(grades[0:2])
print(grades[1:3])
print(grades)
````
