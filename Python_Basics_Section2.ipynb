## 1. Arrays
An array stores **many values** in one variable. The index starts at **0**.


`students[1]` means: go to position 1 and give me the value.

```python
#Arrays
students = ["John", "Jane", "Jack", "Jill"] #Length of the array is 4 #Index: 0, 1, 2, 3 (length - 1)
# print(students)
# print(students[0])
# print(students[1])
# print(students[2])
# print(students[3])

#Time Complexity: O(n) ? (Performance of an algorithm)

#Two accesses: O(1) + O(1) = O(1)
print(students[1]) #O(1)
print(students[2]) #O(1)
```

### Time Complexity (very simple)
It tells us **how the time grows when the data grows**.

Getting a value by its index (`students[1]`) is **one step**, even if the array has a million items. Python goes directly to that position. We call this **O(1)** (constant time).

## 4. String vs Array
A string works like an array: every character has an index.

**Slicing** `[start:stop]` cuts a part. It stops **before** `stop`, and the original is **not changed**.

```python
name = "Amira"
print(name[0])
print(name[-1]) #Last Character
print(name[0:3]) #Substring (from index 0 to index 3)
print(name) # original string not affected by slicing

grades = [10, 20, 30, 40, 50]
print(grades[0:2]) #Slicing (from index 0 to index 2)
print(grades[1:3]) #Slicing (from index 1 to index 3)
print(grades) # original array not modified
```

## 6. Bubble Sort
Sorting means putting the numbers in order (small to big).

**Idea:** compare two neighbors. If the left one is bigger, **swap** them. After each full pass, the biggest number goes to the end.

- Outer loop = the passes
- Inner loop = comparing the neighbors
- `grades[j], grades[j + 1] = grades[j + 1], grades[j]` swaps two values in one line

```python
grades = [10, 5, 6, 20, 15, -3, 5] # => [-3, 5, 5, 6, 10, 15, 20]
print(grades)

#Bubble Sort Algorithm

for i in range(len(grades) - 1):
    for j in range(len(grades) - i - 1):
        if grades[j] > grades[j + 1]:
            grades[j], grades[j + 1] = grades[j + 1], grades[j]

print(grades)
print("Highest Grade:", grades[-1])
print("Lowest Grade:", grades[0])
```

### See every pass (extra)
Same code, but it prints the array after each pass so you can see how the biggest numbers move to the end.

```python
grades = [10, 5, 6, 20, 15, -3, 5]
print("Start     :", grades)

for i in range(len(grades) - 1):
    for j in range(len(grades) - i - 1):
        if grades[j] > grades[j + 1]:
            grades[j], grades[j + 1] = grades[j + 1], grades[j]
    print("After pass", i + 1, ":", grades)
```

## 10. Linear Search
**Idea:** check the items **one by one** from the start until we find the value.

- If we find it: `return i` (the index)
- If we finish the loop and did not find it: `return -1`

Try it: change `search` to `7` and run again.

```python
def linear_search(array, search):
    for i in range(len(array)):
        if array[i] == search:
            return i
    return -1

grades = [3, 5, 12, 10, 4]
search = 12
result = linear_search(grades, search)

if result == -1:
    print("The value " + str(search) + " is not found in the array.")
else:
    print("The value " + str(search) + " is found at index " + str(result))
```

## 11. Binary Search
**Idea:** look in the **middle**. If the value is bigger, throw away the left half. If it is smaller, throw away the right half. Repeat.

**Important:** the array must be **sorted**.

- `left` and `right` are the edges of the part we are searching in
- `mid = (left + right) // 2` is the middle (`//` gives a whole number)
- Found it: return `mid`
- Middle value is too small: search the right half (`left = mid + 1`)
- Middle value is too big: search the left half (`right = mid - 1`)

```python
def binary_search(array, search):
    left = 0
    right = len(array) - 1

    while left <= right:
        mid  = (left + right) // 2
        if array[mid] == search:
            return mid
        elif array[mid] < search:
            left = mid + 1
        else:
            right = mid - 1
    return -1

grades = [10, 12, 14, 15, 17, 19, 20]
search = 20
result = binary_search(grades, search)

if result == -1:
    print("The value " + str(search) + " is not found in the array.")
else:
    print("The value " + str(search) + " is found at index " + str(result))
```

### See every step (extra)
Same search, but it prints `left`, `right` and `mid` in each step.

```python
grades = [10, 12, 14, 15, 17, 19, 20]
search = 20

left = 0
right = len(grades) - 1

while left <= right:
    mid = (left + right) // 2
    print("left =", left, " right =", right, " mid =", mid, " value =", grades[mid])
    if grades[mid] == search:
        print("Found at index", mid)
        break
    elif grades[mid] < search:
        left = mid + 1
    else:
        right = mid - 1
```

## 12. Tuple

## 13. Dictionary (Hash Table)
A dictionary stores data as **key : value** pairs inside `{ }`.

We get a value by its **key** (not by an index): `person['name']`.

```python
person = {'name': 'Ali', 'age': 24, 'city': 'Cairo', 'country': 'Egypt', 'gender': 'Male'}

print(person)
print(person['name'])
print(person['age'])
print(person['city'])
print(person['country'])
print(person['gender'])

person['age'] = 25
print(person['age'])
```

### An array of dictionaries
We can put many dictionaries inside an array. `persons[2]['name']` means: take the item at index 2, then take its `name`.

```python
persons = [
    {'name': 'Ali', 'age': 24, 'city': 'Cairo', 'country': 'Egypt', 'gender': 'Male'},
    {'name': 'Ahmed', 'age': 30, 'city': 'Alexandria', 'country': 'Egypt', 'gender': 'Male'},
    {'name': 'Sara', 'age': 21, 'city': 'Giza', 'country': 'Egypt', 'gender': 'Female'}

]

print(persons[2]['name'])

for i in range(len(persons)):
    print(persons[i]['name'])
```
